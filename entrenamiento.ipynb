```python
import json
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# 1. Cargar Iris
iris = load_iris()
X = iris.data

# 2. Estandarizar las variables
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Entrenar K-means
model = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=20
)

model.fit(X_scaled)

# 4. Preparar los datos para el HTML
data = {
    "features": [
        "Longitud del sépalo",
        "Anchura del sépalo",
        "Longitud del pétalo",
        "Anchura del pétalo"
    ],
    "observations": X.tolist(),
    "observations_scaled": X_scaled.tolist(),
    "labels": model.labels_.tolist(),
    "centroids_scaled": model.cluster_centers_.tolist(),
    "centroids_original": scaler.inverse_transform(
        model.cluster_centers_
    ).tolist(),
    "species": iris.target.tolist(),
    "species_names": iris.target_names.tolist(),
    "mean": scaler.mean_.tolist(),
    "scale": scaler.scale_.tolist()
}

# 5. Exportar a JSON
with open("iris_kmeans.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

print("Archivo JSON generado correctamente")
```
